# Whole-gene assembly from oligopools / OMEGA (template)

Library of independent full-length sequences (orthologs, generative designs, deep multi-mutants), routed to [OMEGA](https://github.com/RomeroLab/omega) for pooled Golden Gate assembly. Fill in the input cell, then run top to bottom. A worked example is in `../tutorials/03-omega-sequence-set.ipynb`.

OMEGA is a separate GPL-3.0 tool; `library_designer` never imports or bundles it. Install it once (see its README) and point `omega_home` at your checkout.

```python
from library_designer import LibrarySpec, CodonOptimizationParams, SequenceSet, OmegaParams

# ===== EDIT: your sequences =====
# A {name: protein} mapping, or load a FASTA with SequenceSet.from_fasta(spec, path).
designs = {
    "design_01": "",   # one-letter protein, no stop codon
    "design_02": "",
}

spec = LibrarySpec(
    name="my_omega_library",
    optimization=CodonOptimizationParams(species="e_coli"),
    platform="pooled",
)
spec
```

## Build and check

```python
lib = SequenceSet(spec, proteins=designs).generate().codon_optimize()
print(lib.summary())
print(lib.check())
```

## Assemble with OMEGA

Needs a separately-installed OMEGA. Returns the oligo order plus per-gene and per-subpool tables.

```python
import os

result = lib.assemble_with_omega(
    OmegaParams(njunctions=40),       # Golden Gate sites per subpool
    omega_home="~/repos/omega",       # your OMEGA checkout, or set OMEGA_HOME
    omega_python=None,                # e.g. "~/miniconda3/envs/omega/bin/python"
)
os.makedirs("out", exist_ok=True)
result.oligos.to_csv("out/oligo_order.csv", index=False)
lib.to_design_specs(f"out/{lib.spec.name}_design_specs.json")

print("wrote files to", os.path.abspath("out"))
for f in sorted(os.listdir("out")):
    print("  ", f)
```

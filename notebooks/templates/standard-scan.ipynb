# Standard library design (template)

Single-substitution scan of a CDS that fits on one synthesis oligo. Fill in the input cell, then run top to bottom. Copy this file into `personal/` for your own work. A worked example is in `../tutorials/01-standard-scan.ipynb`.

```python
from library_designer import LibrarySpec, CodonOptimizationParams, SubstitutionScan

# ===== EDIT =====
spec = LibrarySpec(
    name="my_library",
    protein_sequence="",              # your protein, one-letter, no stop codon
    substitutions=["A"],              # ["A"] alanine scan; list("ACDEFGHIKLMNPQRSTVWY") for full DMS; add "TAG" for amber
    adaptor_5="",                     # 5' flank added to every oligo (cloning / primer sites); "" for none
    adaptor_3="",                     # 3' flank
    optimization=CodonOptimizationParams(species="e_coli", method="use_best_codon"),
    platform="twist_oligo_pools",     # "pooled" | "arrayed" | a synthesis-methods slug
    max_oligo_length=None,            # set to your synthesis length limit to flag over-long constructs
    seed=0,
)
spec
```

## Build and check

```python
lib = SubstitutionScan(spec).generate().codon_optimize()
print(lib.summary())
```

## Export

```python
import os

if lib.failed:
    lib.drop_failed()
lib.export_all("out")   # master CSV, variants.csv, vendor order, design specs, QC plot

print("wrote files to", os.path.abspath("out"))
for f in sorted(os.listdir("out")):
    print("  ", f)
```
